#Complete Pipeline

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

df = pd.read_csv("/content/Jewellery Customer Segmentation Analysis Personas.csv")

# Drop
df.drop(columns=["CustomerID","CustomerAcquisitionDate","CustomerExitDate",
                 "SalesRevenue","CPA","NumberOfTransactions"], inplace=True)

# Boolean
df["IsActive"] = df["IsActive"].astype(int)

# Encoding
df = pd.get_dummies(df, columns=["Gender","Location","ProductType", "PreferredChannel", "Persona"], drop_first=True)

# Drop date columns before filling NaNs or scaling, as they are not numeric
df.drop(columns=["LastPurchaseDate", "SubscriptionStartDate", "SubscriptionEndDate"], inplace=True)

# Missing - now only numeric columns remain (or dummy variables), so median will work
df.fillna(df.median(numeric_only=True), inplace=True)

# Scaling
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df)

# KMeans
kmeans = KMeans(n_clusters=4, random_state=42)
df["Cluster"] = kmeans.fit_predict(X_scaled)

print(df.groupby("Cluster").mean())
```
